```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_excel('1.xlsx')
df.columns = ['x', 'y']

plt.figure(figsize=(8, 8))
plt.scatter(df.x, df.y, s=10)
plt.grid()

xs = np.linspace(4, 12, 100)

plt.figure(figsize=(8, 8))
plt.scatter(df.x, df.y, s=3)
plt.plot(xs, 0.8*xs - 5.3, color='red')     # линия 1
plt.plot(xs, -2*xs + 18.6, color='green')   # линия 2

l1 = df.y > 0.8* df.x - 5.3
l2 = df.y > -2* df.x + 18.6

df['col'] = 0
df.loc[l1 & ~l2, 'col'] = 1
df.loc[~l1 & l2, 'col'] = 2

plt.figure(figsize=(5,5))
plt.scatter(df.x, df.y, c=df.col, s=3)
plt.grid()

def centres(a, b = 0):
    best, bests = None, np.inf
    for x in a:
        if not b:
            s = np.sqrt(((a - x)**2).sum(axis=1)).sum()
        else:
            s = np.abs(a - x).sum(axis=1).sum()
        if s < bests:
            best = x
            bests = s
    return best
c1 = centres(df[df['col']==0][['x', 'y']].to_numpy())
c2 = centres(df[df['col']==1][['x', 'y']].to_numpy())
c3 = centres(df[df['col']==2][['x', 'y']].to_numpy())
centres = np.array([c1, c2, c3])
print(centres)

px = int(centres[:, 0].mean() * 1000)
py = int(centres[:, 1].mean() * 1000)
print(px, py)
plt.figure(figsize=(6, 6))
plt.scatter(df.x, df.y, c=df.col, s=3)                                   # все точки по кластерам
plt.scatter(centres[:, 0], centres[:, 1], color='red', marker='o', s=10)  # центры
plt.grid()
```
